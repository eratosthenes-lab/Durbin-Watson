**Importing libraries**

```python
import numpy as np
import math
from numpy.random import seed
from numpy.random import randn
import itertools as it
from statsmodels.stats.stattools import durbin_watson
import statsmodels.api as sm
```

**Step 1**. A sample of 'n' numbers (regression residuals) obtained using the 64-bit PCG64 pseudorandom number generator (PRNG) - based on the Permuted Congruential Generator family developed by Dr. Melissa O'Neill - and the calculation of its permutations (n!):

```python
sample_size = 50 # n elements of the random sample
rng = np.random.default_rng(seed=1) # 'seed' is used to initialize the pseudo-random number generator
values = rng.standard_normal(sample_size)
sample_permutations = math.factorial(sample_size) # 50!

print(np.format_float_scientific(sample_permutations, precision = 4, exp_digits=3)) # print the number of permutations in scientific notation
print(sample_permutations) # print the number of permutations as an integer
print(values) # print all n elements of the sample
```

**Step 2**. Durbin-Watson d statistic for this sample:

```python
durbin_watson(values)
```

**Step 3**. Sampling of 7 subsets (simulations) from the set of permutations (n!) using PCG64:

```python
# simulation no. 1: 10 permutations
n_permutation_10 = 10
permutations_10 = np.array([rng.permutation(values) for _ in range(n_permutation_10)])

# simulation no. 2: 100 permutations
n_permutation_100 = 100
permutations_100 = np.array([rng.permutation(values) for _ in range(n_permutation_100)])

# simulation no. 3: 1,000 permutations
n_permutation_1000 = 1000
permutations_1000 = np.array([rng.permutation(values) for _ in range(n_permutation_1000)])

# simulation no. 4: 10,000 permutations
n_permutation_10000 = 10000
permutations_10000 = np.array([rng.permutation(values) for _ in range(n_permutation_10000)])

# simulation no. 5: 100,000 permutations
n_permutation_100000 = 100000
permutations_100000 = np.array([rng.permutation(values) for _ in range(n_permutation_100000)])

# simulation no. 6: 1,000,000 permutations
n_permutation_1million = 1000000
permutations_1million = np.array([rng.permutation(values) for _ in range(n_permutation_1million)])

# simulation no. 7: 10,000,000 permutations
n_permutation_10million = 10000000
permutations_10million = np.array([rng.permutation(values) for _ in range(n_permutation_10million)])
```

**Step 4**. For each of the 7 samples (simulations), d is calculated for every permutation:

```python
DW_statistic_values_p_10 = np.apply_along_axis(durbin_watson, axis=1, arr=permutations_10)
DW_statistic_values_p_100 = np.apply_along_axis(durbin_watson, axis=1, arr=permutations_100)
DW_statistic_values_p_1000 = np.apply_along_axis(durbin_watson, axis=1, arr=permutations_1000)
DW_statistic_values_p_10000 = np.apply_along_axis(durbin_watson, axis=1, arr=permutations_10000)
DW_statistic_values_p_100000 = np.apply_along_axis(durbin_watson, axis=1, arr=permutations_100000)
DW_statistic_values_p_1million = np.apply_along_axis(durbin_watson, axis=1, arr=permutations_1million)
DW_statistic_values_p_10million = np.apply_along_axis(durbin_watson, axis=1, arr=permutations_10million)
```

and the statistical measures are obtained:

```python
mean_permutation_10 = np.mean(DW_statistic_values_p_10)
median_permutation_10 = np.median(DW_statistic_values_p_10)
min_permutation_10 = np.min(DW_statistic_values_p_10)
max_permutation_10 = np.max(DW_statistic_values_p_10)
Q1_permutation_10 = np.percentile(DW_statistic_values_p_10, 25)
Q3_permutation_10 = np.percentile(DW_statistic_values_p_10, 75)
sd_permutation_10 = np.std(DW_statistic_values_p_10)

mean_permutation_100 = np.mean(DW_statistic_values_p_100)
median_permutation_100 = np.median(DW_statistic_values_p_100)
min_permutation_100 = np.min(DW_statistic_values_p_100)
max_permutation_100 = np.max(DW_statistic_values_p_100)
Q1_permutation_100 = np.percentile(DW_statistic_values_p_100, 25)
Q3_permutation_100 = np.percentile(DW_statistic_values_p_100, 75)
sd_permutation_100 = np.std(DW_statistic_values_p_100)

mean_permutation_1000 = np.mean(DW_statistic_values_p_1000)
median_permutation_1000 = np.median(DW_statistic_values_p_1000)
min_permutation_1000 = np.min(DW_statistic_values_p_1000)
max_permutation_1000 = np.max(DW_statistic_values_p_1000)
Q1_permutation_1000 = np.percentile(DW_statistic_values_p_1000, 25)
Q3_permutation_1000 = np.percentile(DW_statistic_values_p_1000, 75)
sd_permutation_1000 = np.std(DW_statistic_values_p_1000)

mean_permutation_10000 = np.mean(DW_statistic_values_p_10000)
median_permutation_10000 = np.median(DW_statistic_values_p_10000)
min_permutation_10000 = np.min(DW_statistic_values_p_10000)
max_permutation_10000 = np.max(DW_statistic_values_p_10000)
Q1_permutation_10000 = np.percentile(DW_statistic_values_p_10000, 25)
Q3_permutation_10000 = np.percentile(DW_statistic_values_p_10000, 75)
sd_permutation_10000 = np.std(DW_statistic_values_p_10000)

mean_permutation_100000 = np.mean(DW_statistic_values_p_100000)
median_permutation_100000 = np.median(DW_statistic_values_p_100000)
min_permutation_100000 = np.min(DW_statistic_values_p_100000)
max_permutation_100000 = np.max(DW_statistic_values_p_100000)
Q1_permutation_100000 = np.percentile(DW_statistic_values_p_100000, 25)
Q3_permutation_100000 = np.percentile(DW_statistic_values_p_100000, 75)
sd_permutation_100000 = np.std(DW_statistic_values_p_100000)

mean_permutation_1million = np.mean(DW_statistic_values_p_1million)
median_permutation_1million = np.median(DW_statistic_values_p_1million)
min_permutation_1million = np.min(DW_statistic_values_p_1million)
max_permutation_1million = np.max(DW_statistic_values_p_1million)
Q1_permutation_1million = np.percentile(DW_statistic_values_p_1million, 25)
Q3_permutation_1million = np.percentile(DW_statistic_values_p_1million, 75)
sd_permutation_1million = np.std(DW_statistic_values_p_1million)

mean_permutation_10million = np.mean(DW_statistic_values_p_10million)
median_permutation_10million = np.median(DW_statistic_values_p_10million)
min_permutation_10million = np.min(DW_statistic_values_p_10million)
max_permutation_10million = np.max(DW_statistic_values_p_10million)
Q1_permutation_10million = np.percentile(DW_statistic_values_p_10million, 25)
Q3_permutation_10million = np.percentile(DW_statistic_values_p_10million, 75)
sd_permutation_10million = np.std(DW_statistic_values_p_10million)
```
